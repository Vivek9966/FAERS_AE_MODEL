```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
from pathlib import Path
```

```python
RESULTS_DIR = Path("/home/vivekbisht/Desktop/faers-signal-detection/results_private")
```

```python
enhanced_results = pd.read_csv(RESULTS_DIR / 'semantic_enhanced_validation.csv')
```

```python

enhanced_results['delta_prr'] = (
    enhanced_results['prr'] - enhanced_results['baseline_prr']
)
```

```python
enhanced_results = enhanced_results.sort_values('delta_prr', ascending=False)
mass  = enhanced_results[enhanced_results['delta_prr'] >100]
mod =enhanced_results[(enhanced_results['delta_prr']>=20 )& (enhanced_results['delta_prr'] <= 100)]
mino = enhanced_results[ (enhanced_results['delta_prr'] >0 )& (enhanced_results['delta_prr'] < 20)]
zer0 = enhanced_results[(enhanced_results['delta_prr'] ==0 )]
```

```python
print("SEMANTIC ENHANCEMENT: DETAILED RESULTS BREAKDOWN")
print(f"MASSIVE IMPROVEMENT (>100%): {len(mass)} cases")
for _, row in mass.iterrows():
        print(f"• {row['drug']:20s} + {row['event']:30s}")
        print(f"     Baseline: {row['prr']:>8.2f} → Enhanced: {row['prr']:>8.2f} ({row['delta_prr']:>+7.1f}%)")
        print(f"     Similar events used: {row['neighbors_used']}, Cases: n={row['baseline_n_cases']}")

print(f" MODERATE IMPROVEMENT ([20%,100%]): {len(mass)} cases")
for _, row in mass.iterrows():
        print(f"• {row['drug']:20s} + {row['event']:30s}")
        print(f"     Baseline: {row['prr']:>8.2f} → Enhanced: {row['prr']:>8.2f} ({row['delta_prr']:>+7.1f}%)")
        print(f"     Similar events used: {row['neighbors_used']}, Cases: n={row['baseline_n_cases']}")
print(f" MINOR IMPROVEMENT (<20%): {len(mass)} cases")
for _, row in mino.iterrows():
        print(f"• {row['drug']:20s} + {row['event']:30s}")
        print(f"     Baseline: {row['prr']:>8.2f} → Enhanced: {row['prr']:>8.2f} ({row['delta_prr']:>+7.1f}%)")
        print(f"     Similar events used: {row['neighbors_used']}, Cases: n={row['baseline_n_cases']}")

print(f" NO IMPROVEMENT (0%): {len(mass)} cases")
for _, row in mass.iterrows():
        print(f"• {row['drug']:20s} + {row['event']:30s}")
        print(f"     Baseline: {row['prr']:>8.2f} → Enhanced: {row['prr']:>8.2f} ({row['delta_prr']:>+7.1f}%)")
        print(f"     Similar events used: {row['neighbors_used']}, Cases: n={row['baseline_n_cases']}")
```

```python
bins=[ 0 ,1,10,50,100,500,1000]
labels = ['0-1%','1-10%','10-50%','50-100%','100-500%','500+%']
enhanced_results['improved_category'] = pd.cut(enhanced_results['delta_prr'],bins=bins , labels=labels,include_lowest=True)

print(enhanced_results['improved_category'].value_counts().sort_index())
```
